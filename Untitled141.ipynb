```python
import math

# Ввод длин катетов
a = float(input("Введите длину первого катета: "))
b = float(input("Введите длину второго катета: "))

# Вычисление гипотенузы по теореме Пифагора: c = sqrt(a^2 + b^2)
hypotenuse = math.sqrt(a**2 + b**2)

# Вычисление площади: S = (a * b) / 2
area = (a * b) / 2

# Вывод результатов
print(f"Гипотенуза: {hypotenuse}")
print(f"Площадь: {area}")
```

```python
# Ввод двух чисел
num1 = float(input("Введите первое число: "))
num2 = float(input("Введите второе число: "))

# Проверка условия
if num1 < num2:
    # Если первое меньше второго, вычисляем сумму
    result = num1 + num2
    print(f"Первое число меньше второго. Сумма: {result}")
else:
    # Иначе (если первое больше или равно второму) вычисляем разность
    result = num1 - num2
    print(f"Первое число не меньше второго. Разность: {result}")
```

```python
# Инициализация счетчиков
positive_count = 0
negative_count = 0
zero_count = 0

print("Введите 10 целых чисел:")

for i in range(10):
    num = int(input(f"Число {i + 1}: "))
    
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
    else:
        zero_count += 1

# Вывод результатов
print(f"Положительных чисел: {positive_count}")
print(f"Отрицательных чисел: {negative_count}")
print(f"Нулей: {zero_count}")
```

```python
# Инициализация переменных для хранения максимума и минимума
# Используем None, чтобы правильно обработать первое введенное число
max_num = None
min_num = None

print("Введите 10 целых чисел:")

for i in range(10):
    num = int(input(f"Число {i + 1}: "))
    
    # Для первого числа устанавливаем и максимум, и минимум равными ему
    if max_num is None or num > max_num:
        max_num = num
        
    if min_num is None or num < min_num:
        min_num = num

# Вычисление разницы
difference = max_num - min_num

# Вывод результатов
print(f"Наибольшее число: {max_num}")
print(f"Наименьшее число: {min_num}")
print(f"Разница между наибольшим и наименьшим: {difference}")
```

```python
# Создаем пустой список для хранения чисел
A = []

print("Введите 10 целых чисел:")
for i in range(10):
    num = int(input(f"Число {i + 1}: "))
    A.append(num)

# Находим максимальный элемент в массиве
max_element = max(A)

# Подсчитываем количество элементов, равных максимальному
count_max = 0
for num in A:
    if num == max_element:
        count_max += 1

# Вывод результата
print(f"Массив: {A}")
print(f"Максимальный элемент: {max_element}")
print(f"Количество чисел, равных максимальному: {count_max}")
```

```python
# Инициализация списка и переменных для сумм
A = []
sum_even_indices = 0  # Сумма элементов с четными индексами (0, 2, 4...)
sum_odd_indices = 0   # Сумма элементов с нечетными индексами (1, 3, 5...)

print("Введите 10 целых чисел:")

for i in range(10):
    num = int(input(f"Элемент с индексом {i}: "))
    A.append(num)
    
    # Проверяем четность индекса
    if i % 2 == 0:
        sum_even_indices += num
    else:
        sum_odd_indices += num

# Вычисляем разницу
difference = sum_even_indices - sum_odd_indices

# Вывод результатов
print(f"Массив: {A}")
print(f"Сумма элементов с четными индексами: {sum_even_indices}")
print(f"Сумма элементов с нечетными индексами: {sum_odd_indices}")
```

```python
# Ввод текста от пользователя
text = input("Введите текст: ")

# Инициализация счетчика
count_digits = 0

# Проходим по каждому символу в тексте
for char in text:
    # Проверка, является ли символ цифрой
    if char.isdigit():
        count_digits += 1

# Вывод результата
print(f"Количество арабских цифр в тексте: {count_digits}")
```

```python
import math

def main():
    print("=== Игра: Компьютер угадывает ваше число ===")
    print("Загадайте целое число, а я попробую его отгадать.")
    
    # 1. Получение диапазона от пользователя
    while True:
        try:
            N = int(input("Введите нижнюю границу диапазона (N): "))
            M = int(input("Введите верхнюю границу диапазона (M): "))
            if N > M:
                print("Ошибка: N не может быть больше M. Попробуйте снова.")
            else:
                break
        except ValueError:
            print("Пожалуйста, введите целые числа.")

    print(f"\nОтлично! Ваше число находится в диапазоне [{N}, {M}].")
    print("Отвечайте на мои вопросы:")
    print("'y' или 'д' — если Да")
    print("'n' или 'н' — если Нет")
    print("'q' — если я ошибся и хочу начать заново\n")

    # 2. Инициализация границ поиска
    low = N
    high = M
    attempts = 0

    # 3. Цикл бинарного поиска
    while low <= high:
        attempts += 1
        
        # Находим середину текущего диапазона
        # Используем целочисленное деление
        mid = (low + high) // 2
        
        # Если диапазон схлопнулся в одно число
        if low == high:
            print(f"Ваше число: {low}?")
            confirm = input("Это верно? (y/n): ").lower()
            if confirm in ['y', 'д', 'yes']:
                print(f"Ура! Я угадал число {low} за {attempts} попыток!")
                return
            else:
                print("Похоже, вы ошиблись в ответах ранее, так как других вариантов нет.")
                return

        # Задаем вопрос
        question = f"Ваше число больше {mid}?"
        answer = input(f"{question} (y/n): ").lower()

        if answer in ['q', 'quit', 'выход']:
            print("Игра завершена.")
            return

        if answer in ['y', 'д', 'yes']:
            # Число больше mid, значит нижняя граница сдвигается на mid + 1
            low = mid + 1
        elif answer in ['n', 'н', 'no']:
            # Число не больше mid (значит меньше или равно), верхняя граница становится mid
            high = mid
        else:
            print("Непонятный ответ. Пожалуйста, используйте 'y' (да) или 'n' (нет).")
            attempts -= 1 # Не считаем эту попытку

    print("Что-то пошло не так. Проверьте, честно ли вы отвечали на вопросы!")

if __name__ == "__main__":
    main()
```
